```python
import json
import os
from collections import defaultdict

ipl_folder = "data/ipl_json"
all_match_files = [f for f in os.listdir(ipl_folder) if f.endswith('.json')]

def get_player_franchise_history(player_name):
    """For a given player, find every team they played for, season by season."""
    # Dictionary: season -> set of teams the player appeared for
    history = defaultdict(set)
    
    for filename in all_match_files:
        file_path = os.path.join(ipl_folder, filename)
        with open(file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
        
        season = str(data['info']['season'])
        players_by_team = data['info']['players']  # dict: team_name -> [list of players]
        
        for team_name, team_players in players_by_team.items():
            if player_name in team_players:
                history[season].add(team_name)
    
    return dict(sorted(history.items()))

# Test it on Cummins
print("PJ Cummins career in our IPL dataset:")
cummins_history = get_player_franchise_history('PJ Cummins')
for season, teams in cummins_history.items():
    print(f"  {season}: {', '.join(teams)}")

print("\nRG Sharma career in our IPL dataset:")
rohit_history = get_player_franchise_history('RG Sharma')
for season, teams in rohit_history.items():
    print(f"  {season}: {', '.join(teams)}")
```

```python
# Phase definitions for T20:
# Powerplay = overs 0-5 (first 6 overs, 0-indexed)
# Middle    = overs 6-14 (overs 7-15)
# Death     = overs 15-19 (last 5 overs)

def get_phase(over):
    if over <= 5:
        return 'Powerplay'
    elif over <= 14:
        return 'Middle'
    else:
        return 'Death'

# Group all balls by phase
from collections import defaultdict
phases = defaultdict(list)

for b in all_balls:
    phase = get_phase(b['over'])
    phases[phase].append(b)

# Compute stats for each phase
print("=" * 70)
print("ROHIT vs CUMMINS — PHASE-BY-PHASE BREAKDOWN")
print("=" * 70)
print(f"{'Phase':<12} {'Balls':>6} {'Runs':>6} {'Wkts':>5} {'Avg':>8} {'SR':>8} {'Dot%':>7}")
print("-" * 70)

for phase_name in ['Powerplay', 'Middle', 'Death']:
    balls = phases.get(phase_name, [])
    if not balls:
        print(f"{phase_name:<12} {0:>6} {0:>6} {0:>5} {'N/A':>8} {'N/A':>8} {'N/A':>7}")
        continue
    
    n = len(balls)
    runs = sum(b['runs_batter'] for b in balls)
    wickets = sum(1 for b in balls if b['wicket'])
    dots = sum(1 for b in balls if b['runs_batter'] == 0 and b['runs_extras'] == 0)
    
    avg = runs / wickets if wickets > 0 else float('inf')
    sr = (runs / n) * 100 if n > 0 else 0
    dot_pct = (dots / n) * 100 if n > 0 else 0
    
    avg_str = f"{avg:.1f}" if wickets > 0 else "—"
    print(f"{phase_name:<12} {n:>6} {runs:>6} {wickets:>5} {avg_str:>8} {sr:>7.1f} {dot_pct:>6.1f}%")

print("=" * 70)
```

```python
import json
import os
from collections import defaultdict, Counter

ipl_folder = "data/ipl_json"

def find_matchup_balls(batter_name, bowler_name):
    """
    Scan all IPL match files and return every ball where the given
    bowler delivered to the given batter.
    
    Returns a list of dictionaries, one per ball.
    """
    all_match_files = [f for f in os.listdir(ipl_folder) if f.endswith('.json')]
    
    balls = []
    
    for filename in all_match_files:
        file_path = os.path.join(ipl_folder, filename)
        with open(file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
        
        match_date = data['info']['dates'][0]
        match_season = data['info']['season']
        match_venue = data['info']['venue']
        
        for innings_index, innings in enumerate(data['innings']):
            for over in innings['overs']:
                for delivery in over['deliveries']:
                    if delivery['bowler'] == bowler_name and delivery['batter'] == batter_name:
                        balls.append({
                            'match_id': filename.replace('.json', ''),
                            'date': match_date,
                            'season': match_season,
                            'venue': match_venue,
                            'innings': innings_index + 1,
                            'over': over['over'],
                            'runs_batter': delivery['runs']['batter'],
                            'runs_extras': delivery['runs']['extras'],
                            'runs_total': delivery['runs']['total'],
                            'wicket': 'wickets' in delivery,
                        })
    
    return balls


def get_phase(over):
    """Bucket a T20 over number into a phase."""
    if over <= 5:
        return 'Powerplay'
    elif over <= 14:
        return 'Middle'
    else:
        return 'Death'


def analyze_matchup(batter_name, bowler_name):
    """
    Given a batter and bowler, return a complete matchup analysis:
    headline stats + phase breakdown.
    """
    balls = find_matchup_balls(batter_name, bowler_name)
    
    if not balls:
        return {
            'batter': batter_name,
            'bowler': bowler_name,
            'total_balls': 0,
            'message': 'No deliveries found between these players in our dataset.'
        }
    
    # Headline stats
    total_balls = len(balls)
    total_runs = sum(b['runs_batter'] for b in balls)
    dismissals = sum(1 for b in balls if b['wicket'])
    fours = sum(1 for b in balls if b['runs_batter'] == 4)
    sixes = sum(1 for b in balls if b['runs_batter'] == 6)
    dots = sum(1 for b in balls if b['runs_batter'] == 0 and b['runs_extras'] == 0)
    
    # Phase breakdown
    phase_breakdown = {}
    for phase_name in ['Powerplay', 'Middle', 'Death']:
        phase_balls = [b for b in balls if get_phase(b['over']) == phase_name]
        if phase_balls:
            n = len(phase_balls)
            r = sum(b['runs_batter'] for b in phase_balls)
            w = sum(1 for b in phase_balls if b['wicket'])
            d = sum(1 for b in phase_balls if b['runs_batter'] == 0 and b['runs_extras'] == 0)
            phase_breakdown[phase_name] = {
                'balls': n,
                'runs': r,
                'wickets': w,
                'avg': r / w if w > 0 else None,
                'sr': (r / n) * 100,
                'dot_pct': (d / n) * 100,
            }
        else:
            phase_breakdown[phase_name] = {'balls': 0}
    
    return {
        'batter': batter_name,
        'bowler': bowler_name,
        'total_balls': total_balls,
        'total_runs': total_runs,
        'dismissals': dismissals,
        'avg': total_runs / dismissals if dismissals > 0 else None,
        'sr': (total_runs / total_balls) * 100,
        'fours': fours,
        'sixes': sixes,
        'dots': dots,
        'phase_breakdown': phase_breakdown,
        'matches_played': len(set(b['match_id'] for b in balls)),
    }


# TEST IT — Rohit vs Cummins
result = analyze_matchup('RG Sharma', 'PJ Cummins')

print("=" * 60)
print(f"{result['batter']} vs {result['bowler']}")
print("=" * 60)
print(f"Matches:     {result['matches_played']}")
print(f"Balls:       {result['total_balls']}")
print(f"Runs:        {result['total_runs']}")
print(f"Dismissals:  {result['dismissals']}")
print(f"Average:     {result['avg']:.2f}" if result['avg'] else "Average: Not out")
print(f"Strike rate: {result['sr']:.2f}")
print(f"Fours/Sixes: {result['fours']}/{result['sixes']}")
print()
print("Phase breakdown:")
for phase, stats in result['phase_breakdown'].items():
    if stats['balls'] == 0:
        print(f"  {phase:<12} —  no balls in this phase")
    else:
        avg = f"{stats['avg']:.1f}" if stats['avg'] is not None else "—"
        print(f"  {phase:<12} {stats['balls']:>3} balls | {stats['runs']:>3} runs | {stats['wickets']} wkts | avg {avg:>5} | SR {stats['sr']:>5.1f} | dots {stats['dot_pct']:>4.1f}%")
```

```python
def analyze_batter_overall(batter_name):
    """
    Compute a batter's overall stats in IPL across all bowlers.
    Phase-split. Used as a baseline for matchup comparisons.
    """
    all_match_files = [f for f in os.listdir(ipl_folder) if f.endswith('.json')]
    
    balls = []
    
    for filename in all_match_files:
        file_path = os.path.join(ipl_folder, filename)
        with open(file_path, 'r', encoding='utf-8') as f:
            data = json.load(f)
        
        for innings_index, innings in enumerate(data['innings']):
            for over in innings['overs']:
                for delivery in over['deliveries']:
                    if delivery['batter'] == batter_name:
                        balls.append({
                            'over': over['over'],
                            'runs_batter': delivery['runs']['batter'],
                            'runs_extras': delivery['runs']['extras'],
                            'wicket': 'wickets' in delivery,
                        })
    
    if not balls:
        return None
    
    # Overall stats
    total_balls = len(balls)
    total_runs = sum(b['runs_batter'] for b in balls)
    dismissals = sum(1 for b in balls if b['wicket'])
    
    overall = {
        'balls': total_balls,
        'runs': total_runs,
        'wickets': dismissals,
        'avg': total_runs / dismissals if dismissals > 0 else None,
        'sr': (total_runs / total_balls) * 100,
    }
    
    # Phase-split stats
    phase_stats = {}
    for phase_name in ['Powerplay', 'Middle', 'Death']:
        phase_balls = [b for b in balls if get_phase(b['over']) == phase_name]
        if phase_balls:
            n = len(phase_balls)
            r = sum(b['runs_batter'] for b in phase_balls)
            w = sum(1 for b in phase_balls if b['wicket'])
            phase_stats[phase_name] = {
                'balls': n,
                'runs': r,
                'wickets': w,
                'avg': r / w if w > 0 else None,
                'sr': (r / n) * 100,
            }
    
    return {
        'batter': batter_name,
        'overall': overall,
        'phase_stats': phase_stats,
    }


# Get Rohit's overall baseline
print("Computing Rohit's career baseline... (this scans 1216 matches, takes 10-30 sec)")
rohit_baseline = analyze_batter_overall('RG Sharma')

print()
print("=" * 70)
print(f"ROHIT SHARMA — CAREER IPL BASELINE")
print("=" * 70)
print(f"Career: {rohit_baseline['overall']['balls']} balls, {rohit_baseline['overall']['runs']} runs, "
      f"{rohit_baseline['overall']['wickets']} dismissals")
print(f"Overall average: {rohit_baseline['overall']['avg']:.2f}")
print(f"Overall strike rate: {rohit_baseline['overall']['sr']:.2f}")
print()
print("Phase-split baseline:")
for phase, stats in rohit_baseline['phase_stats'].items():
    avg = f"{stats['avg']:.1f}" if stats['avg'] is not None else "—"
    print(f"  {phase:<12} {stats['balls']:>5} balls | {stats['runs']:>5} runs | "
          f"{stats['wickets']:>3} wkts | avg {avg:>5} | SR {stats['sr']:>5.1f}")
```

```python
def compare_matchup_to_baseline(batter_name, bowler_name):
    """
    Show how a specific bowler matchup compares to the batter's career baseline.
    This is what produces real insight — context, not just stats.
    """
    matchup = analyze_matchup(batter_name, bowler_name)
    baseline = analyze_batter_overall(batter_name)
    
    if matchup['total_balls'] == 0:
        print(f"No matchup data for {batter_name} vs {bowler_name}.")
        return
    
    print("=" * 78)
    print(f"{batter_name} vs {bowler_name} — VS CAREER BASELINE")
    print("=" * 78)
    print(f"Sample: {matchup['total_balls']} balls across {matchup['matches_played']} matches")
    print()
    
    # Header
    print(f"{'Phase':<12} {'Metric':<6} {'vs Bowler':>12} {'Career Avg':>13} {'Difference':>13}")
    print("-" * 78)
    
    for phase in ['Powerplay', 'Middle', 'Death']:
        m = matchup['phase_breakdown'].get(phase, {'balls': 0})
        b = baseline['phase_stats'].get(phase, {'balls': 0})
        
        if m['balls'] == 0:
            print(f"{phase:<12} (no balls in this phase against {bowler_name})")
            continue
        
        # Average
        m_avg = m.get('avg')
        b_avg = b.get('avg')
        m_avg_str = f"{m_avg:.1f}" if m_avg is not None else "—"
        b_avg_str = f"{b_avg:.1f}" if b_avg is not None else "—"
        
        if m_avg is not None and b_avg is not None:
            diff = m_avg - b_avg
            diff_str = f"{diff:+.1f}"
        else:
            diff_str = "—"
        
        print(f"{phase:<12} {'Avg':<6} {m_avg_str:>12} {b_avg_str:>13} {diff_str:>13}")
        
        # Strike rate
        m_sr = m['sr']
        b_sr = b['sr']
        sr_diff = m_sr - b_sr
        print(f"{'':12} {'SR':<6} {m_sr:>12.1f} {b_sr:>13.1f} {sr_diff:>+12.1f}")
        print(f"{'':12} {'Balls':<6} {m['balls']:>12} {b['balls']:>13} {'':>13}")
        print()
    
    print("=" * 78)


# Run the comparison for Rohit vs Cummins
compare_matchup_to_baseline('RG Sharma', 'PJ Cummins')
```
